## LA DENSIDAD  
El término *“densidad”* proviene del campo de la física y la química y alude a la **relación que existe entre la masa de una sustancia (o de un cuerpo) y su volumen.** Se trata de una propiedad intrínseca de la materia, ya que no depende de la cantidad de sustancia que se considere.

### ¿Cuáles son los factores que afectan la determinación de la densidad?
1. **Temperatura:** La temperatura ejerce influencia en el espacio necesario para acomodar los átomos de una molécula en un cierto espacio, porque al vibrar más rápido el espacio entre ellos aumenta.
2. **Burbujas de aire o impurezas:** Una simple burbuja de aire presente en el líquido que se va a medir puede causar una diferencia enorme en su densidad; lo mismo aplica a las impurezas.
3. **Presión atmosférica de un fluido:** Si se encuentra en la Ciudad de México, a 3.930 metros sobre el nivel del mar, la presión atmosférica local será menor que en Río de Janeiro, que está al nivel del mar (0 m). Esto significa que la presión atmosférica está directamente relacionada con la altitud.
    >**Nota:** Salvo que se aplique una presión extremadamente alta, este no es el caso para fines de medición analítica de densidad.

4. **Viscosidad**:La viscosidad describe la resistencia de un líquido al fluir, descrita informalmente como el "espesor" de un fluido, y en principio no tiene una relación directa con la densidad.  


| Método | ¿Influye la viscosidad? | Comentario |
|:---:|:---:|:---:|
| Picnómetro | No | Pero el muestreo y la limpieza son tediosos, y la estabilización de la temperatura es lenta al usar baño de agua. |

#### Densidad o densidad absoluta  

Es la relación entre la masa y el volumen de una sustancia, ya sea sólida, líquida o gaseosa. Se representa por la letra griega rho (𝞺):  

$$
𝞺 = \frac{m}{V}$$
